# Onwordly untrained baselines (Colab)

Runs the same plan as `experiment: baseline` in `.kaggle-run`: builds the frozen evaluation data, then scores each model's untrained greedy answers with the exact parsers and the lenient diagnostic.

Runtime → Change runtime type → **T4 GPU** before running.

```
!git clone https://github.com/HereLiesAz/onwordly.git
%cd onwordly
!pip install -q -e '.[train]'
```

```
MODELS = "Qwen/Qwen2.5-0.5B,Qwen/Qwen2.5-0.5B-Instruct"
CHAT_TEMPLATE = "both"  # no | yes | both (yes/both need a model with a chat template)
PER_SPLIT = 200

plan = f"experiment: baseline\nmodels: {MODELS}\nchat_template: {CHAT_TEMPLATE}\nper_split: {PER_SPLIT}\n"
open('/content/baseline-plan', 'w').write(plan)
print(plan)
```

```
!onwordly-kaggle --plan /content/baseline-plan --output-root /content/results
```
